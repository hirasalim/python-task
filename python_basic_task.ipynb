```python
#task 2

celsius = float(input("Enter temperature in Celsius: "))

fahrenheit = (celsius * 9/5) + 32

print(f"Temperature in Fahrenheit: {fahrenheit}")

if celsius < 10:
    print("Weather: Very Cold")
elif celsius <= 19:
    print("Weather: Cold")
elif celsius <= 29:
    print("Weather: Normal")
else:
    print("Weather: Hot")
```

```python
#task 3

Mobile_Number = input("Enter mobile number: ")

Recharge_amount = int(input("Enter recharge amount: "))

Tax = Recharge_amount * 3 / 100

Final_Balance = Recharge_amount - Tax

print("=================================================")

print(f"Mobile Number: {Mobile_Number}")
print(f"Tax: {Tax}")
print(f"Final Balance: {Final_Balance}")

```

```python
#task 4


total_step = 0

for i in range(7):
    day_step = int(input("Enter your day step: "))
    total_step = total_step + day_step

average = total_step / 7
print("==================================================")
print(f"Total steps:       {total_step}")
print(f"Daily Average steps:  {average}")
if average >5000:
    print("status: active")
else:
    print("need more activity")
```

```python
#task 5

secret_pin = 1234
user_pin = int(input("enter your pin: "))
for i in range(1, 3):
    if user_pin == secret_pin:
        print("Acess Granted")
        break
    else:
        print("Acess denied")
        print(f"no of attempt left {3 - i}")
        user_pin = int(input("enter your pin: "))
if user_pin != secret_pin:
    print("Acess blocked")
```
